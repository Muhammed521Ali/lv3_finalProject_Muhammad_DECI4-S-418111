```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("/content/github_projects.csv")
df['creation'] = pd.to_datetime(df['creation'])
```

```python
df.head()
```

```python
nstarsn = df.sort_values(by="stargazers", ascending=False)
```

```python
stars = nstarsn.head(10)
hjh=stars['stargazers']/1000
```

```python

plt.figure(figsize=(15,15), facecolor='white')
ax = plt.gca()
ax.set_facecolor('#D4BBA1')
Barplot1=plt.bar(stars["owner"],hjh,
    color='#140d1c')
plt.xlabel("owner")
plt.ylabel("stargazers Count(1=1000)")
plt.title("Top 10 repositories by stars")
plt.minorticks_on()
plt.grid(visible=True, which='major', color='#999999', linestyle='-', linewidth=2, alpha=0.5)
plt.grid(visible=True, which='minor', color='#999999', linestyle='--', linewidth=1.2, alpha=0.2)
plt.savefig('Barplot1.png')
plt.show()
```

```python
Lstars = nstarsn.tail(10)
jhj=Lstars['stargazers']/1000
```

```python
plt.figure(figsize=(15,15), facecolor='white')
ax = plt.gca()
ax.set_facecolor('#D4BBA1')
Barplot2=plt.bar(Lstars["owner"],jhj,
    color='#140d1c')
plt.xlabel("owner")
plt.ylabel("stargazers Count(1=1000)")


plt.title("Lower 10 repositories by stars")
plt.minorticks_on()
plt.grid(visible=True, which='major', color='#999999', linestyle='-', linewidth=2, alpha=0.5)
plt.grid(visible=True, which='minor', color='#999999', linestyle='--', linewidth=1.2, alpha=0.2)
plt.savefig('Barplot2.png')
plt.show()
```

```python
nFain=df.sort_values(by="open_issues", ascending=True)
Fai =nFain.tail(10)
Faiv=Fai["open_issues"]/1000
```

```python
plt.figure(figsize=(15,15), facecolor='white')
ax = plt.gca()
ax.set_facecolor('#D4BBA1')
Barplot4=plt.bar(Fai["owner"],Faiv,
    color='#140d1c')
plt.xlabel("owner")
plt.ylabel("open issues Count(1=1000)")
plt.title("Top 10 repositories by open issues")
plt.minorticks_on()
plt.grid(visible=True, which='major', color='#999999', linestyle='-', linewidth=2, alpha=0.5)
plt.grid(visible=True, which='minor', color='#999999', linestyle='--', linewidth=1.2, alpha=0.2)
plt.savefig('Barplot3.png')
plt.show()
```

```python
LFai = nFain.head(10)
kjk=LFai['open_issues']/1000
```

```python
plt.figure(figsize=(15,15), facecolor='white')
ax = plt.gca()
ax.set_facecolor('#D4BBA1')
Barplot3=plt.bar(LFai["owner"],kjk,
    color='#140d1c')
plt.xlabel("owner")
plt.ylabel("open issues Count(1=1000)")
plt.title("Lower 10 repositories by open issues")
plt.minorticks_on()
plt.grid(visible=True, which='major', color='#999999', linestyle='-', linewidth=2, alpha=0.5)
plt.grid(visible=True, which='minor', color='#999999', linestyle='--', linewidth=1.2, alpha=0.2)
plt.savefig('Barplot4.png')
plt.show()
```

```python
Worst = df[df["owner"].isin(Lstars["owner"]) & df["owner"].isin(Fai["owner"])][["owner"]]
Best    = df[df["owner"].isin(stars["owner"]) & df["owner"].isin(LFai["owner"])][["owner"]]
Veryg  = df[df["owner"].isin(stars["owner"])][["owner"]]
Bad=df[df["owner"].isin(Lstars["owner"]) ][["owner"]]
Best=Best.drop_duplicates()
Veryg=Veryg.drop_duplicates()
Bad=Bad.drop_duplicates()
Worst=Worst.drop_duplicates()
print("=== Owners Classification Report ===")
print("\n[Best Owners]:")
print(Best)

print("\n[Very Good Owners]:")
print(Veryg)

print("\n[Bad Owners]:")
print(Bad)

print("\n[Worst Owners]:")
print(Worst)

```
